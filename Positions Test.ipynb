```python
max_date_query = """
SELECT MAX(periodenddate) as max_date
FROM views.dynamicpositions
"""

max_date_df = spark.sql(max_date_query)
max_date = max_date_df.collect()[0]['max_date']

positions_query = f"""
SELECT * 
FROM views.dynamicpositions
WHERE periodenddate = '{max_date}'
"""

positions_df = spark.sql(positions_query)
display(positions_df)
```
